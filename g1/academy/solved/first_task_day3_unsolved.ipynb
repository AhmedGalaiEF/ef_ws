## Task 2 - Command one joint: `move_ll_joint(joint_id, q, dq=0.0, kp=40.0, kd=1.0, tau=0.0, dev_mode=False, ...)`
Commands a single joint while holding every other upper-body joint at its currently observed position, then ramps smoothly to `q` on its own (no manual loop needed). `q`/`dq`/`kp`/`kd`/`tau` are a virtual spring-damper: `kp` pulls toward the target position, `kd` damps velocity, `tau` is a feed-forward torque added on top. The default (`dev_mode=False`) path uses `rt/arm_sdk` and only accepts waist/arm joints (`12-28`) while AI Sport stays the nominal owner.

**Safety:** `kd` (damping) is hard-clamped to `MAX_KD = 10.0` inside the publisher no matter what you pass -- treat `10` as a *ceiling*, not a target. Start much lower (`kd` ~= `1-3`) and only raise it if the joint feels underdamped/oscillatory. `kp` (stiffness) is **not** clamped -- a large `kp` on a joint far from its current position moves it fast and hard, so always read `q` first (Task 1) and command a small delta from it (the example below moves `+/-0.2` to `0.5` rad from the live position, never an arbitrary absolute angle).

## Task 5 - Dex3 hand basics: open / close / release / tactile pressure
- `open_dex3_hand(hand="both", hold_s=0.6, rate_hz=50.0, ramp_s=None)` / `close_dex3_hand(...)` -- ramp to preset open/closed finger targets using gentle gains (`kp=1.5`, `kd=0.1`, `tau=0.03`; fingers are light and don't need arm-scale stiffness).
- `release_fingers(hand="right", hold_s=0.5, rate_hz=50.0, persistent=True)` -- makes the fingers backdrivable (near-zero gain) at their last commanded pose. `persistent=True` (the default) keeps doing this on a background thread until you call `stop_release_fingers()`.
- `get_dex3_tactile_pressures(hand="both")` -- returns per-sensor pressure floats you can use to sense contact/grasp force.

**Safety:** always `open_dex3_hand` before `close_dex3_hand` on an unfamiliar object -- closing onto something already in the palm without a known pressure baseline can over-grip it.
