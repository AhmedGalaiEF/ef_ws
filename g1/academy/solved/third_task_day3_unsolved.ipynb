## Task 2 - `ik_move_ee(hand="right", pose_increment=(dx, dy, dz, droll, dpitch, dyaw), max_speed=0.35, max_dq=0.2, rate_hz=50.0, position_only=False)`
`pose_increment` is a **relative** step in meters/radians from the end effector's *current* pose, not an absolute target -- you call this repeatedly (e.g. from jog buttons) to walk the hand somewhere.

**Safety clamp #1 -- Cartesian step:** regardless of what you pass, each call's step is clamped internally to at most `0.15` m per position axis and `0.6` rad per rotation axis *before* solving. That means the example below, `pose_increment=(1, 0, 1, 0, 0, 0)` -- a full **1 meter** on `x` and `z` -- is silently reduced to the safe `0.15` m step, not sent as literally 1 m.

**Safety clamp #2 -- joint delta:** `max_dq` (default `0.2` rad) caps how far the *solved joint* is allowed to move from its current position in one call, independent of the Cartesian clamp above.

`max_speed` (rad/s) and `rate_hz` control how smoothly (not how far) the ramp to that solved target plays out. If IK fails (target unreachable or singular), nothing moves -- joints stay exactly where they were, so it's always safe to just try a smaller step next.

`position_only=True` ignores `droll`/`dpitch`/`dyaw` and solves position only -- recommended default until you're comfortable with full 6-DOF orientation control.
