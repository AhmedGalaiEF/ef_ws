## Introduction
This task covers `teach()` / `repeat()`: record an arm motion by physically guiding the robot's arm with your hands, then play it back automatically. `teach(name)` makes the arm (and any reachable Dex3 fingers) backdrivable with light damping while the **waist stays fixed** and stiff, and records continuously until you press Enter. `repeat(name)` replays the recorded motion smoothly over `rt/arm_sdk`, blended with AI Sport -- no dev mode involved.

**Known caveat:** occasionally the arm can feel locked/not freely movable right after `teach()` is called, if a previous release/engage hasn't fully settled yet. If that happens, re-run `release_arms()` first and try again -- this is exactly why the last cell in this notebook calls `release_arms()` a second time, as a safety net, even though it was already called once at the start.
