```python
RIGHT_ELBOW = 25
current_elbow = rob.get_joint_position("right_arm.elbow")
if current_elbow is None:
    raise RuntimeError("Could not read right elbow position.")

target_elbow = min(float(current_elbow) - 1, 2.0944)
lift_result = rob.move_upper_body_joint(
    RIGHT_ELBOW,
    target_elbow,
    command_rate_hz=50.0,
    max_speed_rad_s=0.25,
)
```

```python
RIGHT_ELBOW = 24
current_elbow = rob.get_joint_position("right_arm.shoulder_yaw")
if current_elbow is None:
    raise RuntimeError("Could not read right elbow position.")

target_elbow = min(float(current_elbow) - 0.9, 2.0944)
lift_result = rob.move_upper_body_joint(
    RIGHT_ELBOW,
    target_elbow,
    command_rate_hz=50.0,
    max_speed_rad_s=0.25,
)
```
