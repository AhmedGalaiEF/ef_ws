# Introductory robot motion

This notebook creates a `Robot` from `sdk_client`, speaks a warning, then runs the high-level hand wave motion.

Load the local SDK wrapper and choose the DDS interface.

```python
import os
import sys
from pathlib import Path

NOTEBOOK_DIR = Path.cwd().resolve()
MODULES_DIR = NOTEBOOK_DIR.parent
if str(MODULES_DIR) not in sys.path:
    sys.path.insert(0, str(MODULES_DIR))

IFACE = os.environ.get("G1_IFACE", "eth0")
DOMAIN_ID = int(os.environ.get("G1_DOMAIN_ID", "0"))

from sdk_client import Robot

print(f"Configured for iface={IFACE!r}, domain_id={DOMAIN_ID}.")
```

Create the robot client. Sensor startup is disabled because this example only needs speech and a high-level arm action.

```python
robot = Robot(
    iface=IFACE,
    domain_id=DOMAIN_ID,
    safety_boot=False,
    recover_dev_mode_on_init=False,
    auto_start_sensors=False,
)
print("Robot client ready.")
```

Announce the motion before moving.

```python
say_code = robot.say("warning I am about to wave my hand")
print(f"Robot.say returned {say_code}")
```

Run the high-level wave hand motion.

```python
wave_code = robot.face_wave()
print(f"Robot.high_wave returned {wave_code}")
```
